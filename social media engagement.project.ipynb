## Task 2: Data Cleaning and Preprocessing

Data cleaning is the process of identifying and correcting missing values, duplicate records, incorrect data types, and inconsistent data.

It helps improve data quality and makes the dataset suitable for accurate analysis.

### Step 1: Checking Missing Values

The isnull() and isna() functions are used to identify missing values in a dataset.

The sum() function counts the number of missing values in each column.

```python
df.isnull().sum()
```

```python
df.isna().sum()
```

### Checking Missing Values

In Pandas, `isnull()` and `isna()` are used to detect missing values in a dataset.

Both functions perform the same operation and return:
- `True` for missing values.
- `False` for non-missing values.

The `sum()` function is used to count the missing values in each column.

**Note:** Both `isnull()` and `isna()` give the same result.

### Handling Missing Values

Missing values can affect data analysis and produce inaccurate results.

In this step, we identify missing values and handle them using suitable techniques such as:

- `dropna()` – Removes rows or columns containing missing values.
- `fillna()` – Replaces missing values with specified values.
- Median – Fills missing numerical values using the middle value.
- Mode – Fills missing categorical values using the most frequent value.
- Forward Fill – Fills missing values using the previous value.
- Backward Fill – Fills missing values using the next value.

### Missing Value Analysis

In this step, we calculate the percentage of missing values in each column to understand the amount of missing data.

```python
(df.isnull().sum() / len(df)) * 100
```

### Interpretation of Missing Values

The missing value analysis shows that 6 columns contain missing data.

The columns age, gender, likes, comments, shares, and sentiment each have 3% missing values.

The remaining columns do not contain any missing values.

Since only a small percentage of data is missing, suitable imputation techniques can be used to preserve useful records.

```python
df[['age', 'likes', 'comments', 'shares']].median()
```

### Filling Missing Numerical Values Using Median

The median is the middle value of a numerical dataset.

We use the median to fill missing values in the age, likes, comments, and shares columns.

This method helps preserve the dataset without removing useful records.m

```python
df['age'] = df['age'].fillna(df['age'].median())
df['likes'] = df['likes'].fillna(df['likes'].median())
df['comments'] = df['comments'].fillna(df['comments'].median())
df['shares'] = df['shares'].fillna(df['shares'].median())
```

```python
# verify 
df[['age', 'likes', 'comments', 'shares']].isnull().sum()
```

### Filling Missing Categorical Values Using Mode

Mode is the most frequently occurring value in a dataset.

In this step, we use mode to fill missing values in the gender and sentiment columns.

```python
df[['gender', 'sentiment']].mode()
```

```python
df['gender'] = df['gender'].fillna(df['gender'].mode()[0])
df['sentiment'] = df['sentiment'].fillna(df['sentiment'].mode()[0])
```

```python
df[['gender', 'sentiment']].isnull().sum()
```

### Final Verification of Missing Values

After filling numerical missing values using median and categorical missing values using mode, we check the entire dataset to confirm that no missing values remain.

```python
df.isnull().sum()
```

### Checking Duplicate Records

Duplicate records are repeated rows in a dataset. Identifying and removing duplicates helps maintain data accuracy and consistency.

```python
df.duplicated().sum()
```

### Interpretation of Duplicate Records

The duplicate check shows that the dataset contains 0 duplicate rows.

Therefore, no duplicate records need to be removed.

```python
#Remove Duplicates
df = df.drop_duplicates()
```

```python
#verify 

df.duplicated().sum()
```

### Step 2: Log Transformation

In this step, we apply logarithmic transformation to engagement metrics using NumPy's log1p() function.

The log1p() function calculates log(1 + x) and also handles zero values safely.

This helps reduce the impact of large numerical values during analysis.

```python
df['log_likes'] = np.log1p(df['likes'])
df['log_comments'] = np.log1p(df['comments'])
df['log_shares'] = np.log1p(df['shares'])

df[['likes', 'log_likes',
    'comments', 'log_comments',
    'shares', 'log_shares']].head()
```

A line chart shows how engagement changes over time.
